# Lab: Non-Linear Modeling

Linearity assumption is almost always an approximation. We can enhance the least squares by using Ridge, Lasso, principal components regression and other techniques, but that is sometimes not enough. We aim then to be able to relax the linearity assumption while also attempting to maintain interpretability. In particular:

- **Polinomial Regression**
    - We add extra predictors, **raising each of the original predictors to a power**: $x$, $x^2$, $x^3$... 
- **Step functions**
    - We cut the range (the $x$ axis) into $K$ different regions in order to produce a qualitative variable: **We fit each region into a _constant_ function**.
- **Regression splines**, **piecewise polynomials**:
    - A mix between step functions and polinomial regression: We divide the range of $X$ into $K$ different regions, and we fit a polynomial function on each regions. It requires a new constraint: for the polynomials to join smoothly on the boundaries, called _knots_.
- **Smoothing splines**
    - It's similar to Regression splines, but arises from a different situation, as these splines come from minimizing a residual sum of squares criterion, but with a smoothness penalty.
- **Local regression**
    - Not that different to splines, but the regions are allowed to overlap (smoothtly)! 
- **Generalized Additive Models**
    - Extend the previous models for multiple predictors.
    - $y \sim ExponentialFamily(\mu|X)$, where $g(\mu|X) = \beta_0 + f_1(X_1) + f_2(X_2, X3) + \ldots + f_M(X_N)$
        - Esto significa que el valor a obtener, $y$, se puede modelar mediante una distribución exponencial con parámetro $\mu$, condicionada a los valores $X$, y que ese parámetro $\mu$ será el resultado de sumar (o aplicar una relación lineal dada por $g$) el efecto de los distintos predictores, afectados previamente por distintas funciones $f_n$.
    - The **distribution** $y$ should be from the exponential family:
        - normal, binomial, poisson, gamma, inv_gauss, using  ``GAM(distribution='...')``.
    - the **link function** $g$ is the one that should be linear in nature: Once we apply $g$ to the distribution mean, we have a linear prediction.
        - identity, logit, inverse, log, inverse-squared, using ``GAM(link='...')``.
    - the funtional form has an **additive** structure: $\beta_0 + f_1(X_1) + f_2(X_2, X3) + \ldots + f_M(X_N)$, with some terms:
        - `l()` linear terms: for terms like $X_i$
        - `s()` spline terms
        - `f()` factor terms
        - `te()` tensor products
        - `intercept`  


In particular, Polynomial, piecewise-constant and regression splites are special cases of basis functions approach, where the idea is to have at hand a family of transformations to be applied to the variable X (the predictors). Those functions, selected ahead of time (known), would be $b_1(X)$, $b_2(X)$, ..., $b_K(X)$, so that instead of fitting a linear model with $X$, we do it with the $b_k(X)$:
$$
y_i = \beta_0 + \beta_1 b_1(x_i) + \beta_2 b_2(x_i) + \cdot\cdot\cdot + \beta_K b_K(x_i) + \epsilon_i
$$

## Polynomial Regression and Step Functions

Implies using this polynomial function:
$$
y_i = \beta_0 + \beta_1 x_i + \beta_2 x_i^2 + \beta_3 x_i^3 + ... + \beta_d x_i^d + \epsilon_i
$$

As this is just a standard linear prediction, a standard regression can be used, even with Ridge and Lasso constraints, to estimate the coefficients. The predictors would be the original predictors along the powers of those predictors. As the coefficients are now severely correlated, the stability of the process suffers.

**QR orthogonalization** is a mathematical process that transforms a set of correlated features (like polynomial terms $x$,$x^2$,$x^3$, etc) into a new set of uncorrelated (orthogonal) features. This is particularly useful in regression to avoid multicollinearity (when predictors are highly correlated), unstability of the coefficients and low interpretability. QR orthogonalization fixes this by **creating new features that are independent of each other** while preserving the same predictive power.
- Input: A matrix $X$ of raw polynomial terms (e.g., [$x$,$x^2$,$x^3$]:
- QR Decomposition:
    - Breaks $X$ into two matrices, so $X=Q \cdot R$:
        - $Q$ (orthogonal columns, i.e., uncorrelated features)
        - $R$ (upper triangular, used for scaling)
        - Computed using Gram-Schmidt process or Householder reflections.

    
- The new features (columns of $Q$) are orthogonal (uncorrelated).
- The regression problem becomes numerically stable.

### Using scikit-learn vanilla

````python
from scipy.linalg import qr as scipy_qr

def orthogonal_poly(x, degree):
    """
    Replicates ISLP.poly() behavior using sklearn + QR decomposition
    """
    # Standardize first
    x_scaled = (x - x.mean())  # This could have been another transformer, just a scaler.
    
    # Create raw polynomial matrix
    poly = PolynomialFeatures(degree=degree, include_bias=True) # This results in x, x^2, x^3, correlated!
    X_raw = poly.fit_transform(x_scaled.reshape(-1, 1)) # apply the transformer

    # Reshape was needed because the transformers expects a 2D array:
    # Each row: A sample
    # Each column: A feature.
    
    # Orthogonalize using QR decomposition using scipy
    Q, R = scipy_qr(X_raw, mode='economic')
    
    return Q

age_manual = Wage['age'].values
X_poly_manual = orthogonal_poly(age_manual, degree=4)
X_poly_manual_df = pd.DataFrame(X_poly_manual)

M_manual = sm.OLS(y, X_poly_manual_df).fit() # we have now changed the input columns to a set of orthogonal polynomials, and NOW we fit with OLS.
summarize(M_manual)
````

````python
# Fit both models
M_islp = sm.OLS(y, poly_age.transform(Wage)).fit()
M_manual = sm.OLS(y, X_poly_manual_df).fit()

# Compare predictions (should be identical)
predictions_identical = np.allclose(M_islp.predict(), M_manual.predict())
print("Predictions identical:", predictions_identical)
````

Finally, we wish to plot the data and add the fit from the fourth-degree polynomial. As we will make
several similar plots below, we first write a function
to create all the ingredients and produce the plot.  Our function
takes in a model specification (here a basis specified by a
transform), as well as a grid of `age` values. The function
produces a fitted curve as well as 95% confidence bands. By using
an argument for `basis` we can produce and plot the results with several different
transforms, such as the splines we will see shortly. 

````python
def plot_wage_fit(age_df, 
                  basis, # should have a transform() method
                  title):
    
    # transform the X using the transform method received by paramater
    X = basis.transform(Wage) # Wage is the entire dataframe,with X and Y values.
    Xnew = basis.transform(age_df) # and transform the ones we want to predict

    # Fit the OLS model, predict
    M = sm.OLS(y, X).fit() # y is the wage, X is the age.
    preds = M.get_prediction(Xnew)
    bands = preds.conf_int(alpha=0.05) # determine the confidence interval
    
    fig, ax = subplots(figsize=(8,8))
    ax.scatter(age,
               y,
               facecolor='gray',
               alpha=0.5)
    for val, ls in zip([preds.predicted_mean,
                      bands[:,0],
                      bands[:,1]],
                     ['b','r--','r--']):
        ax.plot(age_df.values, val, ls, linewidth=3)
    ax.set_title(title, fontsize=20)
    ax.set_xlabel('Age', fontsize=20)
    ax.set_ylabel('Wage', fontsize=20);
    return ax
````

We include an argument `alpha` to `ax.scatter()`
to add some transparency to the points. This provides a visual indication
of density. Notice the use of the `zip()` function in the
`for` loop above (see Section 2.3.8).
We have three lines to plot, each with different colors and line
types. Here `zip()` conveniently bundles these together as
iterators in the loop. {In `Python`  speak, an "iterator" is an object with a finite number of values, that can be iterated on, as in a loop.}

We now plot the fit of the fourth-degree polynomial using this
function.

````python
plot_wage_fit(age_df, # the values we want to predcit
              poly_age, # the fit model (not transformed: it will transform it inside). It will then apply OLS!
              'Degree-4 Polynomial');
````

### ANOVA use to find the polynomial degree to use

With  polynomial regression we must decide on the degree of
the polynomial to use. Sometimes we just wing it, and decide to use
second or third degree polynomials, simply to obtain a nonlinear fit. But we can
make such a decision in a more systematic way. One way to do this is through **hypothesis
tests**, which we demonstrate here. We now fit a series of models ranging from
linear (degree-one) to degree-five polynomials,
and look to determine the simplest model that is sufficient to
explain the relationship between `wage` and `age`. We use the
`anova_lm()`  function, which performs a series of ANOVA
tests.
An **analysis of
  variance**  or **ANOVA** tests the null
hypothesis that a model $\mathcal{M}_1$ is sufficient to explain the
data against the alternative hypothesis that a more complex model
$\mathcal{M}_2$ is required. The **determination is based on an F-test**.
To perform the test, the models  $\mathcal{M}_1$ and $\mathcal{M}_2$ must be *nested*:
the space spanned by the predictors in $\mathcal{M}_1$ must be a subspace of the
space spanned by the predictors in $\mathcal{M}_2$. In this case, we
fit five different polynomial
models and sequentially compare the simpler model to the more complex
model.

````python
models = [MS([poly('age', degree=d)]) 
          for d in range(1, 6)] # These are, 6 possible models, with up to 6 degrees.
Xs = [model.fit_transform(Wage) for model in models]  # fit all of the models.
anova_lm(*[sm.OLS(y, X_).fit() # apply an ANOVA test. This receives several parameters. As we are sending a list, we use *.
           for X_ in Xs])
````

Notice the `*` in the `anova_lm()` line above. This
function takes a variable number of non-keyword arguments, in this case fitted models.
When these models are provided as a list (as is done here), it must be 
prefixed by `*`.

The p-value comparing the linear `models[0]` to the quadratic
`models[1]` is essentially zero, indicating that a linear
fit is not sufficient, because when considering that `models[0]` is the hypothesis, the p-value is too low, making it highly unlikely for the samples being what they are under that model. Up to this point, `models[1]` would be the adecuate.

Similarly the p-value comparing the quadratic
`models[1]` to the cubic `models[2]` (row 2) is very low (0.0017), so the
quadratic fit is also insufficient.

The p-value comparing the cubic and degree-four polynomials, `models[2]` and `models[3]` (row 3), is
approximately 5%, while the degree-five polynomial `models[4]` seems
unnecessary because its p-value is 0.37 (row 4). Hence, either a cubic or a
quartic polynomial appear to provide a reasonable fit to the data, but
lower- or higher-order models are not justified.

In this case, instead of using the `anova()`  function, we could
have obtained these p-values more succinctly by exploiting the fact
that `poly()`  creates orthogonal polynomials.

````python
summarize(M)
````

Notice that the p-values are the same, and in fact the square of
the  t-statistics are equal to the F-statistics from the
`anova_lm()`  function;  for example:

polynomials, provided the models are nested. For example, we can use
`anova_lm()`  to compare the following three
models, which all have a linear term in `education` and a

````python
models = [MS(['education', poly('age', degree=d)])
          for d in range(1, 4)]
XEs = [model.fit_transform(Wage)
       for model in models]
anova_lm(*[sm.OLS(y, X_).fit() for X_ in XEs])
````

### Use of a polynomial logistic regression model

Next we consider the task of predicting whether an individual earns
more than $250,000 per year. We proceed much as before, except
that first we create the appropriate response vector, and then apply
the `glm()`  function using the binomial family in order
to fit a **polynomial logistic regression model** that uses the polynomial functions of `age` as predictors.

$$
Pr(y_i>250 | x_i) = \frac{exp(\beta_0+\beta_1 x_i + \beta_2 x_i^2 + ... + \beta_d x_i^d)}{1+exp(\beta_0+\beta_1 x_i + \beta_2 x_i^2 + ... + \beta_d x_i^d)}
$$

By picking the corresponding $\beta_i$, and therefore defining a threshold of $Pr(y>250)$ above which we decide for `1` instead of `0` (the turning point), we end up deciding for new predicted values. 

**What is a GLM again?** (from the Ch04-classification-lab.ipynb file)

>An extension for Linear Regression, but for different kinds of response variables and allowing for non-linear relationships between the response and predictors, but linear in the parameters. It models the **mean of the response** using a link function applied to a linear combination of predictors.
>- **Random component**: The response variable as a distribution (Gaussian, Bernoulli, Poisson).
>- **Systematic Component**: The linear predictor, $\eta = \beta_0 + \beta_1 x_1 + \cdot \cdot \cdot \beta_p x_p = \vec{X}\vec{\beta}$.
>- **Link function** $g$, connecting the expected value of the response to the linear predictor $\eta$: $g(E(Y)) = \eta$.
>
>The model won't make the output a linear function of the predictors, but just linear on the parameters $\beta$ _before_ applying the link function. So we _do_ have a linear predictor, $\eta$, but followed by a non-linear function $g$.
>
> In **logistic regressions**, instead of modelling the variable $Y(x)$, it models $P(y \in category)$, which ranges between 0 and 1, by the use on a S-shaped, non-linear transformation.
>
>- **Random component**: It's Bernoulli, given that the response $Y$ is binary (0 and 1), with $P(y=1)=p$ and $P(y=0)=1-p$, with mean $E[Y] = p$. The probability mass function is $f(y;p) = p^y (1-p)^{1-y}$, with $y \in \{0,1\}$. By using `family=Binomial()`, the Binomial log-likelihood is used for estimation instead of least-squares. It also decides to use the corresponding link function `logit` (although it can be overriden).
>- **Systematic component**: The linear predictor is $\eta = \vec{X}\vec{\beta}$, but $\eta$ can't be $p$ because they have different domains (one goes between 0 and 1, and the other one extends or contracts the predictors). We have to use a non-linear function to map this linear predictor into a non-linear function.
>- **Link function**: We need to map $p \in (0,1)$ to $\eta \in (-\infty, \infty)$, so we define the $logit$ function:
>
>$$
logit(p) = \log\left( \frac{p}{1-p}\right)
$$
>so that $logit(p) = \vec{X}\vec{\beta} = \eta$: This is the non-linear function that ends up giving us a linear predictor: the logit is the one modeled as a linear function of the predictors. Solving for $p$ we get the logistic function in terms of $\vec{X}$ and $\vec{\beta}$: 
>$$
p_0 = \frac{1}{1+e^{-\vec{X}\vec{\beta}}}; p_1 = \frac{e^{-\vec{X}\vec{\beta}}}{1+e^{-\vec{X}\vec{\beta}}}
$$ 
>This is the **logistic function**, so $p$ has an S-shape between 0 and 1. **We have to find the _turning point_ where the probability density changes. This is changed by the $\beta_i$ values to be found.**
>
>This $p$ is the one we would use for expressing the probability of the output we actually got, as it's still a set of Bernoullies we are talking about. The parameter $p$ in the Bernoulli is the one we aim to end up calculating.
>The `sm.GLM()`  function fits *generalized linear models*, a class of
models that includes logistic regression.  Alternatively,
the function `sm.Logit()` fits a logistic regression
model directly. The syntax of
`sm.GLM()` is similar to that of `sm.OLS()`, except
that we must pass in the argument `family=sm.families.Binomial()`
in order to tell `statsmodels` to run a logistic regression rather than some other
type of generalized linear model.

>```mermaid
flowchart LR
     Statsmodels --> ols("OLS(samples,correct_values)") --> lr("Linear Regression")
     Statsmodels --> glm["GLM(samples,correct_values)"] --"family=sm.families.Binomial()" --> logr["Logistic Regression"]
     glm --"family=sm.families.Gaussian()"--> lr
     glm --"family=sm.families.Poisson()"--> Counting
```


````python
X = poly_age.transform(Wage) # Remember, poly_age = MS([poly('age', degree=4)]).fit(Wage)
high_earn = Wage['high_earn'] = y > 250 # shorthand, added a new column with True/False.

# We will use a generalized linear model
glm = sm.GLM(y > 250, # response, whether it's a high earner or not.
             X, # matrix of predictors
             family=sm.families.Binomial()) # the family for the random component, the distribution for Y. in this case Binomial count: Add up 0 or 1.
B = glm.fit()
summarize(B)
````

## Step functions

This means that we are converting the continuous variable into an ordered categorical variable: the X value fits into a certain bin and for being on that bin, the response variable is the constant value corresponding to that bin. The constant value is usually the mean. The usually _miss the action_.

## Splines

### Piecewise Polynomials
- Involves fitting seprate low-degree polynomials over different regions of X. So the coefficients $\beta_0$, $\beta_1$, $\beta_2$ are different in different parts of the range. The points where the coefficients change are called knots.
- Each of the polynomial functions can be fitted using least squares, but we add the contraint that **the fitted curve must be continuous**, and that the first and second derivatives should also be continuous on the knots.  In general, continuity in derivatives is requested to be up to degree $d-1$ for polynomials of degree $d$.

In general, a cubic (d=3) spline can be modeled using basis functions and fit using least squares:
$$
y_i = \beta_0 + \beta_1 b_1(x_i) + \beta_2 b_2 (x_i) + \cdot \cdot \cdot \beta_{K+3} b_{K+3}(x_i) + \epsilon_i
$$
Where the extra 3 parameters (therefore $K+3$ are due to the reduction in degrees of freedom due to the imposed continuities on the knots).
The basis functions are undefined up to this point.

One way to represent polynomials is to start with the basis for cubic polynomials ($1, x, x^2, x^3$) and  add one truncated power basis function per knot, defined as
$$
h(x,\eta) = (x-\eta)_+^3 = (x-\eta)^3 \text{ if } x>\eta \text{ else } 0
$$

where $\eta$ is the knot. It can be shown that adding a term $\beta_4h(x, \eta)$ leads to a discontinuity only in the third derivative, and therefore we want to use in total $3+K$ predictors of the form: $X, X^2, X^3, h(X,\eta_1),h(X,\eta_2),\cdot\cdot\cdot,h(X,\eta_K)$, which ends up **requiring estimating $K+4$ regression coefficients**.

- In general, splines produces more stable results by adding flexibility by means of using more plots instead of using polynomials of higher degrees.

**The problem**: In general, splines have a high variance at the outer range of the predictors, so sometimes a **natural spline** is used, in which a new boundary constraint is added: the function is required to be linear at the boundary (the region where X is smaller than the smallest knot and larger than the largest knot).

**Where to put the knots**
Ideally, more knots should be placed where the function should change more rapidly, but in practice the knots are placed somewhat uniformly.

**How many knots to use?**
In generla we use cross-validation to define that.

In order to fit regression splines, we use transforms
from the `ISLP` package. The actual spline
evaluation functions are in the `scipy.interpolate` package;
we have simply wrapped them as transforms
similar to `Poly()` and `PCA()`.

In Section 7.4, we saw
that regression splines can be fit by constructing an appropriate
matrix of basis functions.  The `BSpline()`  function generates the
entire matrix of basis functions for splines with the specified set of
knots. By default, the B-splines produced are cubic. To change the degree, use
the argument `degree`.

````python
bsp = BSpline(internal_knots=[25,40,60], intercept=True)
bsp
````

````python
bs_ = bsp.fit(age)
print('bs_.knots_:', bs_.knots_) # the knots have been calculated, including the first and last
````

````python
# We actually do apply the pipeline to the data. This will calculate everything for the data itself.
bs_age = bs_.transform(age)
bs_age.shape
bs_age
````

This results in a seven-column matrix, which is what is expected for a cubic-spline basis with 3 interior knots ($K+4$).

For a certain row, the values obtained in each column is the value of that basis function $h(x,\eta)$ for that $x$. Therefore the value of the predicted $y$ will be each of those values for the basis functions, multiplied by the corresponding $\beta_k$ that the model learns.

## GAMs: Generalized Additive Models
Predicting $Y$ with more than one predictor $X_1, ..., X_p$, by allowing non-linear functions on each predictor, but keeping additivity (a different $f_j$ for each $X_j$). This means that the methods seen previously can be used as building blocks for an additive model.
- Allows us to fit non-linear $f_j$ to each $X_j$, modeling then non-linear relationships.
- Each $X_j$ affects the model individually if all other predictors are fixed.
- Important interactions can be missed because the model is forced to be additive.

Given a linear regression model:
$$
y_i = \beta_0 + \beta_1 x_{i1} + \beta_2 x_{i2} + \cdot\cdot\cdot + \beta_p x_{ip} + \epsilon_i
$$
If we were to expect to be able to allow for non-linear relationships, we could eplace $\beta_j x_{ij}$ by a non-linear function $f_i(x_{ij})$, so that
$$
y_i = \beta_0 + f_1(x_{i1}) + f_2(x_{i2}) + \cdot\cdot\cdot + f_p(x_{ip}) + \epsilon_i
$$

For example, we may say
$$
\text{wage} = \beta_0 + f_1(\text{year}) + f_2(\text{age}) +  f_2(\text{education}) + \epsilon
$$

and fit `year` and `age` with smoothing splines, or local regression, or polynomial regression, while fitting `education` (which is a categorical variable) with a different constant for each level using dummy variables.

- Can be used for classification problems ($Y$ being qualitative) with a small twist, using the logistic regression model extended to allow for non-linear relationships so that

$$
\log\left(\frac{p(X)}{1-p(X)}\right) = \beta_0 + f_1(X_1) + f_2(X_2) + \cdot\cdot\cdot + f_p(X_p)
$$
for example, with $f_2$ a smoothing spline of 5 degrees od freedom and $f_3$ a step function:
$$
\log\left(\frac{p(X)}{1-p(X)}\right) = \beta_0 + \beta_1 \times \text{year} + f_2(\text{age}) + f_3(\text{education})
$$
where
$$
p(X) = Pr(\text{wage}>250 | \text{year, age, education})
$$

## Smoothing Splines and GAMs

We want to find a function $g(x)$ that fits the data well (makes $RSS = \sum_{i=1}^{n} (y_i - g(x_i))^2$ very small), but not too well in order to avoid overfitting, so we expect it to be smooth. We may want then to find $g(x)$, the smoothind spline, that minimizes the following expression, that includes a regularization term with a $\lambda$ tuning parameter:

$$
\sum_{i=1}^{n} (y_i - g(x_i))^2 + \lambda \int g''(t)^2 dt
$$

The regularization includes the integral along the whole domain of the second derivative of $g(t)$, and is a way of measuring roughness. The function that minimizes the expression has some special characteristics: It is a piecewise polynomial with knots at the unique values of $x_1$, ..., $x_n$ and it has continuous first and second derivative at each knot. **Least Squates can't be used!**. It also is linear outside the extreme knots, so **the function that minimizes the expression is a natural cubit spline with knots at $x_1, x_2, ..., x_n$**. There will be a knot at each training observation, so we don't have to determine the position of the knots. We instead have to find the value of $\lambda$. In particular, using LOOCV allows us to use a formula to get the RSS for all cases.

A **smoothing spline** is a special case of a GAM (Generalized Additive Model) with squared-error loss
and a single feature. To fit GAMs in `Python` we will use the
`pygam` package which can be installed via `pip install pygam`. The
estimator `LinearGAM()` uses squared-error loss.
The GAM is specified by associating each column
of a model matrix with a particular smoothing operation:
`s` for smoothing spline; `l` for linear, and `f` for factor or categorical variables.
The argument `0` passed to `s` below indicates that this smoother will
apply to the first column of a feature matrix. Below, we pass it a
matrix with a single column: `X_age`. The argument `lam` is the penalty parameter $\lambda$ as discussed in Section 7.5.2.

````python
X_age = np.asarray(age).reshape((-1,1)) # make it a matrix
gam = LinearGAM(s_gam(0, # apply the smoother to the first column of the feature matrix (we will then pass it the column itself, X_age)
                     lam=0.6)) # penalty parameter lambda
gam.fit(X_age, y)
````

The `pygam` library generally expects a matrix of features so we reshape `age` to be a matrix (a two-dimensional array) instead
of a vector (i.e. a one-dimensional array). The `-1` in the call to the `reshape()` method tells `numpy` to impute the
size of that dimension based on the remaining entries of the shape tuple.
 
Let’s investigate how the fit changes with the smoothing parameter `lam`.
The function `np.logspace()` is similar to `np.linspace()` but spaces points
evenly on the log-scale. Below we vary `lam` from $10^{-2}$ to $10^6$.

````python
fig, ax = subplots(figsize=(8,8))
ax.scatter(age, y, facecolor='gray', alpha=0.5)
for lam in np.logspace(-2, 6, 5):
    gam = LinearGAM(s_gam(0, lam=lam)).fit(X_age, y)
    ax.plot(age_grid,
            gam.predict(age_grid),
            label='{:.1e}'.format(lam),
            linewidth=3)
ax.set_xlabel('Age', fontsize=20)
ax.set_ylabel('Wage', fontsize=20);
ax.legend(title='$\lambda$');
````

The `pygam` package can perform a search for an optimal smoothing parameter.

````python
gam_opt = gam.gridsearch(X_age, y)
gam_opt.summary()
````

````python
ax.plot(age_grid,
        gam_opt.predict(age_grid),
        label='Grid search',
        linewidth=4)
ax.legend()
fig
````

Alternatively, we can fix the degrees of freedom of the smoothing
spline using a function included in the `ISLP.pygam` package. Below we
find a value of $\lambda$ that gives us roughly four degrees of
freedom. We note here that these degrees of freedom include the
unpenalized intercept and linear term of the smoothing spline, hence there are at least two
degrees of freedom.

````python
age_term = gam.terms[0]
lam_4 = approx_lam(X_age, age_term, 4)
age_term.lam = lam_4
degrees_of_freedom(X_age, age_term)
````

as the desired degrees of freedom plus one to account for the fact that these smoothing

### Additive Models with Several Terms
The strength of generalized additive models lies in their ability to fit multivariate regression models with more flexibility than linear models. We demonstrate two approaches: the first in a more manual fashion using natural splines and piecewise constant functions, and the second  using the `pygam` package and smoothing splines.

We now fit a GAM by hand to predict
`wage` using natural spline functions of `year` and `age`,
treating `education` as a qualitative predictor, as in (7.16).
Since this is just a big linear regression model
using an appropriate choice of basis functions, we can simply do this
using the `sm.OLS()`  function.

We will build the model matrix in a more manual fashion here, since we wish
to access the pieces separately when constructing partial dependence plots.

````python
ns_age = NaturalSpline(df=4).fit(age) # We fit the age with a natural spline

ns_year = NaturalSpline(df=5).fit(Wage['year']) # And the year with another natural spline

# So we transform the predictors
Xs = [ns_age.transform(age), # 4 columns
      ns_year.transform(Wage['year']), # 5 columns
      pd.get_dummies(Wage['education']).values] # the qualitative predictor is obtained by using get_dummies, 5 columns

# Stack the columns horizontally so the predictor values are the columns.
X_bh = np.hstack(Xs)

# And we use old plain OLS, because we have transformed the predictors beforehand
gam_bh = sm.OLS(y, X_bh).fit()
````

Here the function `NaturalSpline()` is the workhorse supporting
the `ns()` helper function.  We chose to use all columns of the
indicator matrix for the categorical variable `education`, making an intercept redundant.
Finally, we stacked the three component matrices horizontally to form the model matrix `X_bh`. 

```mermaid
flowchart LR
p1[pred_1] -->|transform| tp1[transformed_predictor_1] --> ols[OLS] --> predy[Predicted Y]
p2[pred_2] --> ols
p3[pred_3] -->|transform| tp3[transformed_predictor_3] -->ols
```

We now show how to construct partial dependence plots for each of the terms in our rudimentary GAM. We can do this by hand,
given grids for `age` and `year`.
 We simply predict with new $X$ matrices, fixing all but one of the features at a time.

````python
# Let's create a what-if dataset:
age_grid = np.linspace(age.min(), # create the age grid, evenly spaced
                       age.max(),
                       100)


X_age_bh = X_bh.copy()[:100] # keep only the first 100 rows
X_age_bh[:] = X_bh[:].mean(0)[None,:] # replace all rows with the mean of the original predictors -> all variables are fixed to their average value! We want to isolate the effect of age only.
X_age_bh[:,:4] = ns_age.transform(age_grid) # the first 4 columns are the spline basis for age, so we replace that with the transformed age_grid values.

# With the what-if dataset, let's understand how the values are to be predicted.
preds = gam_bh.get_prediction(X_age_bh) # obtain predicted values for the just-made dataset using the model created before
bounds_age = preds.conf_int(alpha=0.05) # we will plot confidence intervals.
partial_age = preds.predicted_mean

# Let's create the partial dependences, centered at 0, so the Y axis show how much it deviates from the mean contribution.
center = partial_age.mean()
partial_age -= center
bounds_age -= center
fig, ax = subplots(figsize=(8,8))
ax.plot(age_grid, partial_age, 'b', linewidth=3)
ax.plot(age_grid, bounds_age[:,0], 'r--', linewidth=3)
ax.plot(age_grid, bounds_age[:,1], 'r--', linewidth=3)
ax.set_xlabel('Age')
ax.set_ylabel('Effect on wage')
ax.set_title('Partial dependence of age on wage', fontsize=20);
````

Let's explain in some detail what we did above. The idea is to create a new prediction matrix, where all but the columns belonging to `age` are constant (and set to  their training-data means). The four columns for `age` are filled in with the natural spline basis evaluated at the 100 values in `age_grid`.

* We made a grid of length 100 in `age`, and created a matrix `X_age_bh` with 100 rows and the same number of columns as `X_bh`.
* We replaced every row of this matrix with the column means of the original.
* We then replace just the first four columns representing `age` with the natural spline basis computed at the values in `age_grid`. 

The remaining steps should by now be familiar.

We also look at the effect of `year` on `wage`; the process is the same.

````python
year_grid = np.linspace(2003, 2009, 100)
year_grid = np.linspace(Wage['year'].min(),
                        Wage['year'].max(),
                        100)
X_year_bh = X_bh.copy()[:100]
X_year_bh[:] = X_bh[:].mean(0)[None,:]
X_year_bh[:,4:9] = ns_year.transform(year_grid)
preds = gam_bh.get_prediction(X_year_bh)
bounds_year = preds.conf_int(alpha=0.05)
partial_year = preds.predicted_mean
center = partial_year.mean()
partial_year -= center
bounds_year -= center
fig, ax = subplots(figsize=(8,8))
ax.plot(year_grid, partial_year, 'b', linewidth=3)
ax.plot(year_grid, bounds_year[:,0], 'r--', linewidth=3)
ax.plot(year_grid, bounds_year[:,1], 'r--', linewidth=3)
ax.set_xlabel('Year')
ax.set_ylabel('Effect on wage')
ax.set_title('Partial dependence of year on wage', fontsize=20);
````

We now fit the model (7.16)  using smoothing splines rather
than natural splines.  All of the
terms in  (7.16)  are fit simultaneously, taking each other
into account to explain the response. The `pygam` package only works with matrices, so we must convert
the categorical series `education` to its array representation, which can be found

The two `s_gam()` terms result in smoothing spline fits, and use a default value for $\lambda$  (`lam=0.6`), which is somewhat arbitrary. For the categorical term `education`, specified using a `f_gam()` term,  we specify `lam=0` to avoid any shrinkage.
We produce the partial dependence plot in `age` to see the effect of these choices.

The values for the plot
are generated by the `pygam` package. We provide a `plot_gam()`

Note that updating `age_term.lam` above updates it in `gam_full.terms[0]` as well! Likewise for `year_term.lam`.

Repeating the plot for `age`, we see that it is much smoother.

### ANOVA Tests for Additive Models
In all of our models, the function of `year` looks rather linear. We can
perform a series of ANOVA tests in order to determine which of these
three models is best: a GAM that excludes `year` ($\mathcal{M}_1$),
a GAM that uses a linear function of `year` ($\mathcal{M}_2$), or a
GAM that uses a spline function of `year` ($\mathcal{M}_3$).

````python
gam_0 = LinearGAM(age_term + f_gam(2, lam=0))
gam_0.fit(Xgam, y)
gam_linear = LinearGAM(age_term +
                       l_gam(1, lam=0) +
                       f_gam(2, lam=0))
gam_linear.fit(Xgam, y)
````

Notice our use of `age_term` in the expressions above. We do this because
earlier we set the value for `lam` in this term to achieve four degrees of freedom.

To directly assess the effect of `year` we run an ANOVA on the
three models fit above.

````python
anova_gam(gam_0, gam_linear, gam_full)
````

 We find that there is compelling evidence that a GAM with a linear
function in `year` is better than a GAM that does not include
`year` at all ($p$-value= 0.002). However, there is no
evidence that a non-linear function of `year` is needed
($p$-value=0.435).  In other words, based on the results of this
ANOVA, $\mathcal{M}_2$ is preferred.

We can repeat the same process for `age` as well. We see there is very clear evidence that
a non-linear term is required for `age`.

````python
gam_0 = LinearGAM(year_term +
                  f_gam(2, lam=0))
gam_linear = LinearGAM(l_gam(0, lam=0) +
                       year_term +
                       f_gam(2, lam=0))
gam_0.fit(Xgam, y)
gam_linear.fit(Xgam, y)
anova_gam(gam_0, gam_linear, gam_full)
````

## Local Regression
We illustrate the use of local regression using  the `lowess()` 
function from `sm.nonparametric`. Some implementations of
GAMs allow terms to be local regression operators; this is not the case in `pygam`.

Here we fit local linear regression models using spans of 0.2
and 0.5; that is, each neighborhood consists of 20% or 50% of
the observations. As expected, using a span of 0.5 is smoother than 0.2.

````python
lowess = sm.nonparametric.lowess
fig, ax = subplots(figsize=(8,8))
ax.scatter(age, y, facecolor='gray', alpha=0.5)
for span in [0.2, 0.5]:
    fitted = lowess(y,
                    age,
                    frac=span,
                    xvals=age_grid)
    ax.plot(age_grid,
            fitted,
            label='{:.1f}'.format(span),
            linewidth=4)
ax.set_xlabel('Age', fontsize=20)
ax.set_ylabel('Wage', fontsize=20);
ax.legend(title='span', fontsize=15);
````
